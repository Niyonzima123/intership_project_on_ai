```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# ==========================================
# 1. DATA LOADING & CLEANING
# ==========================================

# Load the dataset from the CSV file into a pandas DataFrame
data = pd.read_csv("Book1.csv")

# Remove rows where every single column is empty (fixes the "Input contains NaN" error)
data = data.dropna(how='all')

# Convert the categorical string targets ('PASS' / 'FAIL') into numeric values (1 / 0)
# This prevents scikit-learn from throwing errors when handling the target variable
data['Result'] = data['Result'].map({'PASS': 1, 'FAIL': 0})

# ==========================================
# 2. FEATURE SELECTION & SPLITTING
# ==========================================

# Separate features (X) from the target label (y)
# We drop 'Student' (non-numeric ID) and 'Result' (the target column) using axis=1
X = data.drop(['Student', 'Result'], axis=1)
y = data['Result']

# Split the dataset into training (70%) and testing (30%) sets
# 'random_state=42' ensures the split is reproducible every time you run the code
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# ==========================================
# 3. MODEL TRAINING & EVALUATION
# ==========================================

# Initialize the Decision Tree Classifier, limiting the tree depth to 3 levels to avoid overfitting
model = DecisionTreeClassifier(max_depth=3)

# Train (fit) the decision tree model using the numerical training data
model.fit(X_train, y_train)

# Use the trained model to make predictions on the unseen testing data features
y_pred = model.predict(X_test)

# Calculate and display the model's performance accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.2f}")

# ==========================================
# 4. PREDICTING ON NEW DATA
# ==========================================

# Define a new student's data as a 2D array matching the format of X:
# [Study Hours, Attendance %, Prev Score, Assignments]
# Example: 7 Study Hours, 50% Attendance, Prev Score of 5, 4 Assignments completed
new_student = [[7, 50, 5, 4]]

# Use the trained model to predict the outcome for this new student
prediction = model.predict(new_student)

# Decode the numerical prediction back into a human-readable text output
output_label = "PASS" if prediction[0] == 1 else "FAIL"
print(f"Prediction for new student data {new_student}: {output_label}")
```
